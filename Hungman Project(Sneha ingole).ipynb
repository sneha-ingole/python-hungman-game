```python
import random
word_bank = {
    'easy':{
        'Animals': ['cat', 'dog', 'bat'],
        'Fruits': ['apple', 'pear', 'kiwi']
    },
    'medium': {
        'Technology': ['python', 'laptop', 'gadget'],
        'Colors': ['purple', 'orange', 'voilet']

    },
    'hard': {
        'Programming': ['algorithm', 'debugging', 'inheritance'],
        'Sports': ['basketball', 'cricket', 'badminton']
    }

}

#Hungman ASCII stages
stages = [
    """\n  \n  \n  \n  \n  \n=========""",
    """\n  0\n  \n  \n  \n=========""",
    """\n  0\n  |\n  \n  \n=========""",
    """\n  0\n  /|\n  \n  \n=========""",
    """\n  0\n  /|\\\n  \n  \n=========""",
    """\n  0\n  /|\\\n  / \n=========""",
    """\n  0\n  /|\\\n  / \\\n========="""
]

#Function to select word
def select_word(difficulty, category):
    return random.choice(word_bank[difficulty][category])

#Function to initialize game state
def initialize_game(word):
    return {
        'word': word,
        'hidden_word': ['_'] * len(word),
        'guessed_letters': set(),
        'attempts_left': len(stages) - 1,
        'score': 0
    }

#Process guess
def process_guess(state, guess):
    word = state['word']
    hidden = state['hidden_word']
    guessed = state['guessed_letters']

    if guess in guessed:
        return "Already guessed!"

    guessed.add(guess)

    if guess in word:
        for i in range(len(word)):
            if word[i] == guess:
                hidden[i] = guess
        state['score'] += 10
        return "Correct!"
    else:
        state['attempts_left'] -= 1
        state['score'] -= 2
        return "Wrong!"

#Display game state
def display_game(state):
    print("\n" + stages[len (stages) - 1 - state['attempts_left']])
    print("Word:", ' '.join(state['hidden_word']))
    print(f"Atttempts left: {state['attempts_left']}")
    print(f"Guessed letters: {', '.join(state['guessed_letters'])}")
    print(f"Score: {state['score']}")

#Main game function
def play_hungman():
    print("🎮 Welcome to Enhanced hangamn!\n")
    difficulty = input("Choose difficulty (easy/medium/hard): ").lower()
    category = input(f"Choose category ({'/'.join(word_bank[difficulty].keys())}): ").title()

    word = select_word(difficulty, category)
    state = initialize_game(word)
    
    while state['attempts_left'] > 0:
        display_game(state)
        guess = input("Guess a letter: ").lower()
        if not guess.isalpha() or len(guess) != 1:
            print("please enter a single alphabet.")
            continue
        message = process_guess(state, guess)
        print(message)

        if '_' not in state['hidden_word']:
            print("\n🎉 You won!")
            print(f"The word was: {state['word']}")
            print(f"Final Score: {state['score']}")
            return

        print("\n💀 Game Over!")
        print(f"The word was: {state['word']}")
        print(f"Final Score: {state['score']}")

#Start the game
if __name__ == "__main__":
    play_hungman()
    

        
    
        
```
